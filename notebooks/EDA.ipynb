# Exploratory Data Analysis (EDA) - AQI Prediction for Lahore

This notebook performs exploratory data analysis on the historical AQI and weather data to understand patterns, correlations, and trends.

## 1. Load Data and Initial Exploration

```python
# Cell 1 - Data load
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/historical_backfill.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])
print(df.shape)
df.head()
```

## 2. Check for Missing Values

```python
# Cell 2 - Missing values check
print(df.isnull().sum())
```

## 3. PM2.5 Time Series Analysis

```python
# Cell 3 - AQI time series
plt.figure(figsize=(14, 4))
plt.plot(df["timestamp"], df["pm25"])
plt.title("PM2.5 over time - Lahore")
plt.xlabel("Date")
plt.ylabel("PM2.5")
plt.tight_layout()
plt.show()
```

## 4. Feature Correlation Analysis

```python
# Cell 4 - Correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation")
plt.show()
```

## 5. Hour-wise PM2.5 Pattern Analysis

```python
# Cell 5 - Hour-wise AQI pattern
df["hour"] = df["timestamp"].dt.hour
hourly_avg = df.groupby("hour")["pm25"].mean()
plt.figure(figsize=(10, 4))
hourly_avg.plot(kind="bar", color="steelblue")
plt.title("Average PM2.5 by Hour of Day")
plt.xlabel("Hour")
plt.ylabel("PM2.5")
plt.show()
```

## Summary

This EDA notebook provides insights into:
- Data completeness and missing values
- PM2.5 trends over time for Lahore
- Correlations between air quality and weather features
- Hourly patterns in air pollution levels

Key findings will help with feature selection and model building in the next steps.
